```python
# ============================================================
# Customer Churn Prediction
# Notebook 02 - Data Preparation
# ============================================================

import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)

print("Libraries imported successfully.")
```

```python
# ============================================================
# Load Dataset
# ============================================================

df = pd.read_csv("../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv")

print("Dataset loaded successfully.")
print(f"Shape : {df.shape}")
```

```python
# ============================================================
# Dataset Overview
# ============================================================

display(df.head())

print("\nShape :", df.shape)

print("\nData Types")
display(df.dtypes)
```

```python
# ============================================================
# Missing Values
# ============================================================

missing = pd.DataFrame({
    "Missing Count": df.isna().sum(),
    "Missing %": (df.isna().sum() / len(df) * 100).round(2)
})

missing = missing[missing["Missing Count"] > 0]

missing
```

```python
# ============================================================
# Inspect TotalCharges
# ============================================================

print(df["TotalCharges"].dtype)

df["TotalCharges"].head(10)
```

```python
# ============================================================
# Count Blank Strings
# ============================================================

(df["TotalCharges"] == " ").sum()
```

```python
# ============================================================
# Replace Blank Strings with NaN
# ============================================================

df["TotalCharges"] = df["TotalCharges"].replace(" ", np.nan)

print("Blank strings replaced with NaN.")
```

```python
# ============================================================
# Verify Missing Values
# ============================================================

df["TotalCharges"].isna().sum()
```

```python
# ============================================================
# Convert TotalCharges to Numeric
# ============================================================

df["TotalCharges"] = pd.to_numeric(df["TotalCharges"])

print(df["TotalCharges"].dtype)
```

```python
# ============================================================
# Investigate Missing TotalCharges
# ============================================================

df[df["TotalCharges"].isna()]
```

```python
# ============================================================
# Remove Missing TotalCharges
# ============================================================

df = df.dropna(subset=["TotalCharges"]).reset_index(drop=True)

print("New Shape:", df.shape)
print("Missing TotalCharges:", df["TotalCharges"].isna().sum())
```

```python
# ============================================================
# Duplicate Records
# ============================================================

duplicates = df.duplicated().sum()

print(f"Duplicate Rows : {duplicates}")
```

```python
# ============================================================
# Duplicate Customer IDs
# ============================================================

duplicate_customers = df["customerID"].duplicated().sum()

print(f"Duplicate Customer IDs : {duplicate_customers}")
```

```python
# ============================================================
# Categorical Columns
# ============================================================

categorical_columns = df.select_dtypes(include="object").columns

print(f"Total Categorical Columns : {len(categorical_columns)}\n")

for column in categorical_columns:
    print("=" * 60)
    print(column)
    print("=" * 60)
    print(df[column].value_counts())
    print()
```

```python
# ============================================================
# Save Clean Dataset
# ============================================================

df.to_csv("../data/processed/clean_telco_churn.csv", index=False)

print("Clean dataset saved successfully.")
print(df.shape)
```
